```python
import re
import unicodedata

def normalize_text(text):
    if isinstance(text, list):
        return [normalize_text(x) for x in text]

    text = unicodedata.normalize("NFKC", text)
    text = re.sub(r"\s+", " ", text)  # replace weird spacing with one space
    text = text.strip()

    return text

hc3_clean["question"] = hc3_clean["question"].apply(normalize_text)
hc3_clean["human_answers_normalized"] = hc3_clean["human_answers"].apply(normalize_text)
hc3_clean["chatgpt_answers_normalized"] = hc3_clean["chatgpt_answers"].apply(normalize_text)
```

Drop duplicates

```python
hc3_split = hc3_clean.drop_duplicates(subset=["question"]).copy()
```

**Train/ Test/ Validation Split**

breakdown summary:
- 80% train
- 10% test
- 10% validation

```python
from sklearn.model_selection import train_test_split
```

```python
#first split: 80% train and 20% other
train_questions, temp_questions = train_test_split(hc3_split, test_size=0.20, random_state=42, stratify=hc3_split["source"])
```

```python
#split that remaining 20% evenly: 10% validation and 10% test
val_questions, test_questions = train_test_split(temp_questions, test_size=0.50, random_state=42, stratify=temp_questions["source"])
```

```python
# print results to see number in each split
print("train questions:", len(train_questions))
print("validation questions:", len(val_questions))
print("test questions:", len(test_questions))
```

```python
def create_answer_dataset(df):

  # Human written answers
  human = df[["question", "source", "human_answers_normalized"]].explode("human_answers_normalized").copy()
  human = human.rename(columns={"human_answers_normalized": "text"})
  human["label"] = 0

  #AI generated answers
  ai = df[["question", "source", "chatgpt_answers_normalized"]].explode("chatgpt_answers_normalized").copy()
  ai = ai.rename(columns={"chatgpt_answers_normalized": "text"})
  ai["label"] = 1

  # Combine
  combined = pd.concat( [human, ai], ignore_index=True )

  return combined
```

```python
train_df = create_answer_dataset(train_questions)
val_df = create_answer_dataset(val_questions)
test_df = create_answer_dataset(test_questions)
```

Convert each split into answer level data

```python
#answer level split sizes
print("Train:", train_df.shape)
print("Validation:", val_df.shape)
print("Test:", test_df.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=50000, ngram_range=(1, 2), sublinear_tf=True)
X_train = tfidf.fit_transform(train_df["text"])
X_val = tfidf.transform(val_df["text"])
X_test = tfidf.transform(test_df["text"])
y_train, y_val, y_test = train_df["label"], val_df["label"], test_df["label"]
print(X_train.shape, X_val.shape, X_test.shape)
```

```python
df = train_df.copy()
words = df["text"].str.split()
df["word_count"] = words.str.len()
df["unique_word_ratio"] = words.apply(lambda w: len(set(x.lower() for x in w)) / len(w) if w else 0)
df["punct_per_100_words"] = df["text"].str.count(r"[^\w\s]") / df["word_count"].clip(lower=1) * 100
df["avg_sentence_len"] = df["word_count"] / df["text"].str.count(r"[.!?]+").clip(lower=1)
summary = df.groupby("label")[["word_count", "unique_word_ratio", "punct_per_100_words", "avg_sentence_len"]].mean().round(2)
summary.index = ["Human", "ChatGPT"]
print(summary)
```

## Milestone 2: Linear SVM

**Plan:** (1) train on the train set, (2) use the validation set to pick the best setting `C`, (3) score once on the untouched test set.

### Step 1: Train and tune `C` on the validation set

**What is `C`?** It controls how strictly the SVM must classify every training example correctly:
- **Small C (e.g. 0.01):** the model tolerates some training mistakes and prefers a wide, simple margin. Risk: underfitting.
- **Large C (e.g. 10):** the model tries hard to get every training example right, even weird ones. Risk: overfitting.

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

best_C, best_acc = None, 0
for C in [0.01, 0.1, 1, 10]:
    svm = LinearSVC(C=C, random_state=42)
    svm.fit(X_train, y_train)                      # learn the dividing line from training data
    acc = accuracy_score(y_val, svm.predict(X_val))  # check it on data it hasn't seen
    print(f"C={C}: val accuracy = {acc:.4f}")
    if acc > best_acc:
        best_C, best_acc = C, acc

# retrain with the winning C
svm = LinearSVC(C=best_C, random_state=42).fit(X_train, y_train)
print("Best C:", best_C)
```

### Step 2: Evaluate once on the test set

- **Precision (ChatGPT):** of the answers the model *called* ChatGPT, what fraction really were.
- **Recall (ChatGPT):** of the answers that *really were* ChatGPT, what fraction the model caught.
- **F1:** a single number blending precision and recall (high only if both are high).
- **Accuracy:** overall fraction correct.
- **Confusion matrix:** rows = true class, columns = predicted class. The diagonal is correct predictions. Off-diagonal cells are mistakes: human text flagged as AI (false positives) and AI text that slipped through as human (false negatives).

```python
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

y_pred = svm.predict(X_test)
print(classification_report(y_test, y_pred, target_names=["Human", "ChatGPT"]))
ConfusionMatrixDisplay(confusion_matrix(y_test, y_pred),
                       display_labels=["Human", "ChatGPT"]).plot()
plt.show()
```

### Step 3 (optional): Which words does the SVM rely on?

A linear SVM gives every word a **weight** (`svm.coef_`). Large positive weight = pushes the prediction toward ChatGPT. Large negative weight = pushes toward Human. Sorting the weights shows which words and phrases the model finds most telling, which is useful for the interpretation section of the report.

```python
import numpy as np

feature_names = np.array(tfidf.get_feature_names_out())
order = np.argsort(svm.coef_[0])                  # sorted from most human-like to most ChatGPT-like
print("Most human-like:  ", list(feature_names[order[:15]]))
print("Most ChatGPT-like:", list(feature_names[order[-15:]]))
```

## Linear SVM Results

**Setup.** TF-IDF features (50,000 unigrams and bigrams, fit on train only) fed into scikit-learn's `LinearSVC`. The split is by question (65,548 train / 8,213 validation / 8,202 test answers), so no question appears in more than one split.

### Tuning `C` (validation accuracy)

| C | Validation accuracy |
|---|---|
| 0.01 | 0.9485 |
| 0.1 | 0.9825 |
| **1** | **0.9870** |
| 10 | 0.9855 |

Accuracy rises sharply from `C=0.01` to `C=1`: with a tiny `C` the model is too constrained (underfitting). Past `C=1` it stops improving and slips slightly at `C=10`, the start of overfitting. The curve is flat near the top, so the result is not sensitive to the exact `C`. We chose `C=1`.

### Test set performance (run once, after `C` was fixed)

| Class | Precision | Recall | F1 | Support |
|---|---|---|---|---|
| Human | 0.99 | 1.00 | 0.99 | 5,560 |
| ChatGPT | 0.99 | 0.98 | 0.98 | 2,642 |
| **Overall accuracy** | | | **0.99** | 8,202 |

Confusion matrix (rows = true, columns = predicted):

| | Predicted Human | Predicted ChatGPT |
|---|---|---|
| **True Human** | 5,536 | 24 |
| **True ChatGPT** | 59 | 2,583 |

- **Accuracy is about 99% (8,119 of 8,202 correct).** Test accuracy matches validation accuracy (0.987), which suggests choosing `C` on validation did not overfit.
- **Only 24 human answers (0.4%) were wrongly flagged as AI.** This is the costly mistake in practice, since it accuses a real person of using AI, and it is rare.
- **59 ChatGPT answers (2.2%) slipped through as human.** The model is slightly better at recognizing humans than at catching AI, which is expected because there are about twice as many human examples (5,560 vs 2,642) in the data.
- Because the classes are imbalanced (about 68% human), accuracy alone could look good by always guessing "human" (68%). The per-class precision and recall show the model is genuinely good at both classes.

### What the model learned (top weighted features)

- **Most ChatGPT-like:** `help`, `imagine you`, `helps to`, `it is`, `important`, `important to`, `sure`, `such as`, `including`, `does that`. These are the polite, explanatory, hedged phrasings typical of an assistant ("It is important to...", "Sure!", "Does that help?").
- **Most human-like:** `etc`, `basically`, `can not`, `thus`, `either`, `edit`, `you`, `url_0`. These are casual, forum-style words (`edit` as in "Edit: thanks for the replies").
